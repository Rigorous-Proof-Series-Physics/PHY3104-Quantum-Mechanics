<!-- block:start definition label=linear-vector-space -->
### Linear Vector Space

A linear vector space $X$ over a field $F$ is a collection of vectors

$$
|V\rangle, |W\rangle, \dots \in X
$$

equipped with:

1. Vector addition

$$
X \times X \to X
$$

2. Scalar multiplication

$$
F \times X \to X
$$

satisfying the vector space axioms.
<!-- block:end -->

<!-- block:start axiom label=vector-space-closure -->
### Closure

For all

$$
|V\rangle, |W\rangle \in X,
$$

the sum

$$
|V\rangle + |W\rangle
$$

belongs to $X$.
<!-- block:end -->

<!-- block:start axiom label=vector-space-distributivity -->
### Distributivity

For all scalars $a,b\in F$ and vectors $|V\rangle,|W\rangle\in X$,

$$
a(|V\rangle+|W\rangle)
=
a|V\rangle+a|W\rangle,
$$

and

$$
(a+b)|V\rangle
=
a|V\rangle+b|V\rangle.
$$
<!-- block:end -->

<!-- block:start axiom label=vector-space-commutativity -->
### Commutativity

For all vectors

$$
|V\rangle, |W\rangle \in X,
$$

one has

$$
|V\rangle + |W\rangle
=
|W\rangle + |V\rangle.
$$
<!-- block:end -->

<!-- block:start axiom label=vector-space-associativity -->
### Associativity

Scalar multiplication satisfies

$$
a(b|V\rangle)
=
(ab)|V\rangle,
$$

and vector addition satisfies

$$
|V\rangle + (|W\rangle + |Z\rangle)
=
(|V\rangle + |W\rangle) + |Z\rangle.
$$
<!-- block:end -->

<!-- block:start axiom label=vector-space-identity -->
### Identity

There exists a null vector

$$
|0\rangle
$$

such that

$$
|V\rangle + |0\rangle
=
|V\rangle,
$$

and scalar multiplication satisfies

$$
1|V\rangle
=
|V\rangle.
$$
<!-- block:end -->

<!-- block:start axiom label=additive-inverse -->
### Additive Inverse

For every vector

$$
|V\rangle,
$$

there exists a vector

$$
-|V\rangle
$$

such that

$$
|V\rangle + (-|V\rangle)
=
|0\rangle.
$$
<!-- block:end -->

<!-- block:start definition label=linear-independence -->
### Linear Independence

A set of vectors

$$
|i\rangle
$$

is linearly independent when the only solution to

$$
\sum_i a_i |i\rangle
=
|0\rangle
$$

is

$$
a_i = 0
\qquad
\forall i.
$$
<!-- block:end -->

<!-- block:start definition label=dimension-of-vector-space -->
### Dimension of a Vector Space

A vector space $X$ has dimension $N$ if there exists a maximal set of

$$
N
$$

linearly independent vectors.
<!-- block:end -->

<!-- block:start definition label=basis-of-vector-space -->
### Basis

A basis of a vector space $X$ is a set of linearly independent vectors that spans the entire space.

Any vector

$$
|V\rangle \in X
$$

can be expanded as

$$
|V\rangle
=
\sum_i v_i |i\rangle.
$$
<!-- block:end -->

<!-- block:start definition label=inner-product-space -->
### Inner Product Space

An inner product space is a vector space $X$ equipped with an inner product

$$
\langle V|W\rangle
$$

satisfying:

1. Conjugate symmetry

$$
\langle V|W\rangle
=
\langle W|V\rangle^*
$$

2. Positive definiteness

$$
\langle V|V\rangle \ge 0
$$

with equality iff

$$
|V\rangle = |0\rangle
$$

3. Linearity in the ket

$$
\langle V|
(a|W\rangle+b|Z\rangle)
=
a\langle V|W\rangle
+
b\langle V|Z\rangle.
$$
<!-- block:end -->

<!-- block:start definition label=orthogonal-vectors -->
### Orthogonal Vectors

Two vectors are orthogonal if

$$
\langle V|W\rangle
=
0.
$$
<!-- block:end -->

<!-- block:start definition label=norm-of-vector -->
### Norm of a Vector

The norm of a vector is defined by

$$
|V|
=
\sqrt{\langle V|V\rangle}.
$$
<!-- block:end -->

<!-- block:start definition label=normalized-vector -->
### Normalized Vector

A vector is normalized when

$$
|V| = 1.
$$
<!-- block:end -->

<!-- block:start definition label=orthonormal-basis -->
### Orthonormal Basis

An orthonormal basis is a complete set of vectors satisfying

$$
\langle n|m\rangle
=
\delta_{nm}.
$$
<!-- block:end -->

<!-- block:start definition label=dual-space -->
### Dual Space

The dual space $X^*$ consists of linear functionals acting on vectors of $X$.

Vectors in the dual space are denoted by bras

$$
\langle V|.
$$
<!-- block:end -->

<!-- block:start definition label=hermitian-conjugate-of-ket -->
### Hermitian Conjugate of a Ket

The Hermitian conjugate operation maps a ket to a bra:

$$
(|V\rangle)^\dagger
=
\langle V|.
$$
<!-- block:end -->

<!-- block:start definition label=hermitian-conjugate-of-bra -->
### Hermitian Conjugate of a Bra

The Hermitian conjugate operation maps a bra to a ket:

$$
(\langle V|)^\dagger
=
|V\rangle.
$$
<!-- block:end -->

<!-- block:start proposition label=double-hermitian-conjugation -->
### Double Hermitian Conjugation

For every ket,

$$
(|V\rangle)^{\dagger\dagger}
=
|V\rangle.
$$
<!-- block:end -->

<!-- block:start proposition label=antilinearity-hermitian-conjugation -->
### Antilinearity of Hermitian Conjugation

For complex scalars $a,b$,

$$
(a|V\rangle+b|W\rangle)^\dagger
=
a^* \langle V|
+
b^* \langle W|.
$$
<!-- block:end -->

<!-- block:start definition label=gram-schmidt-procedure -->
### Gram–Schmidt Procedure

Given linearly independent vectors

$$
|I\rangle, |II\rangle, |III\rangle,\dots,
$$

an orthonormal basis can be constructed recursively by subtracting projections and normalizing.
<!-- block:end -->

<!-- block:start theorem label=schwarz-inequality -->
### Schwarz Inequality

For any vectors

$$
|V\rangle,
|W\rangle,
$$

one has

$$
|\langle V|W\rangle|^2
\le
\langle V|V\rangle
\langle W|W\rangle.
$$

<!-- proof:start -->
Define

$$
|Z\rangle
=
|V\rangle
-
\frac{\langle W|V\rangle}{\langle W|W\rangle}
|W\rangle.
$$

Since

$$
\langle Z|Z\rangle \ge 0,
$$

expanding the inner product yields

$$
\langle V|V\rangle
-
\frac{
|\langle V|W\rangle|^2
}{
\langle W|W\rangle
}
\ge 0.
$$

Multiplying by

$$
\langle W|W\rangle
$$

gives

$$
|\langle V|W\rangle|^2
\le
\langle V|V\rangle
\langle W|W\rangle.
$$
<!-- proof:end -->
<!-- block:end -->

<!-- chapter:start title="Operators" -->

<!-- block:start definition label=operator -->
### Operator

An operator is a mapping

$$
\Omega : X \to X
$$

such that

$$
\Omega |V\rangle
=
|V'\rangle.
$$
<!-- block:end -->

<!-- block:start definition label=linear-operator -->
### Linear Operator

An operator $L$ is linear if

$$
L(\alpha|V\rangle+\beta|W\rangle)
=
\alpha L|V\rangle
+
\beta L|W\rangle.
$$
<!-- block:end -->

<!-- block:start definition label=antilinear-operator -->
### Antilinear Operator

An operator $A$ is antilinear if

$$
A(\alpha|V\rangle+\beta|W\rangle)
=
\alpha^* A|V\rangle
+
\beta^* A|W\rangle.
$$
<!-- block:end -->

<!-- block:start definition label=operator-commutator -->
### Commutator

The commutator of two operators is defined by

$$
[\Lambda,\Omega]
=
\Lambda\Omega
-
\Omega\Lambda.
$$
<!-- block:end -->

<!-- block:start definition label=operator-anticommutator -->
### Anticommutator

The anticommutator of two operators is defined by

$$
\{\Lambda,\Omega\}
=
\Lambda\Omega
+
\Omega\Lambda.
$$
<!-- block:end -->

<!-- block:start definition label=outer-product -->
### Outer Product

Given vectors

$$
|V\rangle,
|W\rangle,
$$

the outer product is the operator

$$
|V\rangle\langle W|
$$

defined by

$$
(|V\rangle\langle W|)|\psi\rangle
=
|V\rangle \langle W|\psi\rangle.
$$
<!-- block:end -->

<!-- block:start proposition label=resolution-of-identity -->
### Resolution of the Identity

For an orthonormal basis

$$
|n\rangle,
$$

the identity operator satisfies

$$
\mathbb I
=
\sum_n |n\rangle \langle n|.
$$
<!-- block:end -->

<!-- block:start definition label=hermitian-conjugate-operator -->
### Hermitian Conjugate of an Operator

The Hermitian conjugate of an operator $\Omega$ is defined by

$$
\langle V|\Omega^\dagger|W\rangle
=
\langle W|\Omega|V\rangle^*.
$$
<!-- block:end -->

<!-- block:start definition label=hermitian-operator -->
### Hermitian Operator

An operator is Hermitian if

$$
\Omega^\dagger
=
\Omega.
$$
<!-- block:end -->

<!-- block:start definition label=antihermitian-operator -->
### Anti-Hermitian Operator

An operator is anti-Hermitian if

$$
\Omega^\dagger
=
-\Omega.
$$
<!-- block:end -->

<!-- block:start definition label=unitary-operator -->
### Unitary Operator

An operator is unitary if

$$
\Omega\Omega^\dagger
=
\Omega^\dagger\Omega
=
\mathbb I.
$$

Equivalently,

$$
\Omega^{-1}
=
\Omega^\dagger.
$$
<!-- block:end -->

<!-- block:start definition label=eigenket -->
### Eigenket

A ket

$$
|u\rangle
$$

is an eigenket of an operator $\Omega$ with eigenvalue $a$ if

$$
\Omega |u\rangle
=
a |u\rangle.
$$
<!-- block:end -->

<!-- block:start definition label=spectrum-of-operator -->
### Spectrum of an Operator

The spectrum of an operator is the set of all its eigenvalues.
<!-- block:end -->

<!-- block:start theorem label=hermitian-operator-real-eigenvalues -->
### Hermitian Operators Have Real Eigenvalues

Let

$$
\Omega^\dagger = \Omega
$$

and suppose

$$
\Omega |u\rangle = a |u\rangle.
$$

Then the eigenvalue

$$
a
$$

is real.

<!-- proof:start -->
Starting from the eigenvalue equation,

$$
\Omega |u\rangle = a |u\rangle,
$$

take the expectation value with respect to

$$
\langle u|.
$$

Then

$$
\langle u|\Omega|u\rangle
=
a \langle u|u\rangle.
$$

Since $\Omega$ is Hermitian,

$$
\langle u|\Omega|u\rangle
=
\langle u|\Omega|u\rangle^*.
$$

Therefore,

$$
a\langle u|u\rangle
=
a^*\langle u|u\rangle.
$$

Hence,

$$
(a-a^*)\langle u|u\rangle = 0.
$$

Because

$$
\langle u|u\rangle > 0,
$$

it follows that

$$
a=a^*.
$$

Therefore the eigenvalue is real.
<!-- proof:end -->
<!-- block:end -->

<!-- block:start theorem label=orthogonality-of-hermitian-eigenspaces -->
### Orthogonality of Eigenspaces of Hermitian Operators

Let

$$
\Omega^\dagger = \Omega
$$

and suppose

$$
\Omega |u\rangle = a |u\rangle,
$$

$$
\Omega |u'\rangle = a' |u'\rangle,
$$

with

$$
a \ne a'.
$$

Then

$$
\langle u'|u\rangle = 0.
$$

<!-- proof:start -->
Using the first eigenvalue equation,

$$
\langle u'|\Omega|u\rangle
=
a\langle u'|u\rangle.
$$

Using the second eigenvalue equation,

$$
\langle u|\Omega|u'\rangle
=
a'\langle u|u'\rangle.
$$

Since $\Omega$ is Hermitian,

$$
\langle u|\Omega|u'\rangle
=
\langle u'|\Omega|u\rangle^*.
$$

Therefore,

$$
a'\langle u|u'\rangle
=
a^*\langle u|u'\rangle.
$$

Because the eigenvalues of Hermitian operators are real,

$$
a^* = a.
$$

Hence,

$$
(a-a')\langle u'|u\rangle = 0.
$$

Since

$$
a \ne a',
$$

it follows that

$$
\langle u'|u\rangle = 0.
$$

Therefore the eigenvectors are orthogonal.
<!-- proof:end -->
<!-- block:end -->
