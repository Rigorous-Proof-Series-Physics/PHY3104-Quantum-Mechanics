<!-- block:start postulate label=quantum-state-postulate -->
### Quantum State

At each instant $t$, the state of a physical system is represented by a ket

$$
|\psi\rangle
$$

belonging to a space of states called a Hilbert space.
<!-- block:end -->

<!-- block:start postulate label=observable-operator-postulate -->
### Observables as Operators

Every observable physical quantity is represented by an operator acting on the Hilbert space of states.
<!-- block:end -->

<!-- block:start postulate label=unitary-time-evolution-postulate -->
### Unitary Time Evolution

The time evolution of an isolated quantum system preserves normalization.

Therefore, the evolution is described by a unitary operator

$$
\hat U(t,t_0)
$$

such that

$$
|\psi(t)\rangle
=
\hat U(t,t_0)|\psi(t_0)\rangle.
$$
<!-- block:end -->

<!-- block:start postulate label=measurement-spectrum-postulate -->
### Measurement Outcomes

The only possible outcomes of a measurement of an observable $A$ are the eigenvalues of the operator

$$
\hat A.
$$
<!-- block:end -->

<!-- block:start postulate label=born-rule-postulate -->
### Born Rule

If the system is in the state

$$
|\psi\rangle
$$

and the observable $A$ possesses eigenstates

$$
|a_n\rangle,
$$

then the probability of obtaining the eigenvalue

$$
a_n
$$

is

$$
P(a_n)
=
|\langle a_n|\psi\rangle|^2.
$$
<!-- block:end -->

<!-- block:start postulate label=wavefunction-collapse-postulate -->
### Projection Postulate

Immediately after a measurement yielding the eigenvalue

$$
a_n,
$$

the system collapses to the normalized eigenstate

$$
|a_n\rangle.
$$
<!-- block:end -->

<!-- block:start definition label=linear-vector-space -->
### Linear Vector Space

A linear vector space $X$ over a field $F$ is a collection of vectors

$$
|V\rangle, |W\rangle,\dots \in X
$$

equipped with:

1. Vector addition

$$
X\times X \to X
$$

defined by

$$
|V\rangle + |W\rangle,
$$

2. Scalar multiplication

$$
F\times X \to X
$$

defined by

$$
a|V\rangle,
\qquad a\in F,
$$

satisfying the vector space axioms.
<!-- block:end -->

<!-- block:start axiom label=vector-space-closure -->
### Closure

For all

$$
|V\rangle, |W\rangle \in X,
$$

the sum

$$
|V\rangle + |W\rangle
$$

also belongs to $X$.
<!-- block:end -->
