<!-- subsection:start title="Linear Vector Spaces" -->

<!-- block:start definition label=linear-vector-space -->
### Linear Vector Space

A linear vector space $X$ over a field $F$ is a collection of vectors

$$
|V\rangle, |W\rangle, \dots \in X
$$

equipped with:

1. Vector addition

$$
X \times X \to X
$$

2. Scalar multiplication

$$
F \times X \to X
$$

satisfying the vector space axioms.
<!-- block:end -->

<!-- block:start axiom label=vector-space-closure -->
### Closure

For all

$$
|V\rangle, |W\rangle \in X,
$$

the sum

$$
|V\rangle + |W\rangle
$$

belongs to $X$.
<!-- block:end -->

<!-- block:start axiom label=vector-space-distributivity -->
### Distributivity

For all scalars $a,b\in F$ and vectors $|V\rangle,|W\rangle\in X$,

$$
a(|V\rangle+|W\rangle)
=
a|V\rangle+a|W\rangle,
$$

and

$$
(a+b)|V\rangle
=
a|V\rangle+b|V\rangle.
$$
<!-- block:end -->

<!-- block:start axiom label=vector-space-commutativity -->
### Commutativity

For all vectors

$$
|V\rangle, |W\rangle \in X,
$$

one has

$$
|V\rangle + |W\rangle
=
|W\rangle + |V\rangle.
$$
<!-- block:end -->

<!-- block:start axiom label=vector-space-associativity -->
### Associativity

Scalar multiplication satisfies

$$
a(b|V\rangle)
=
(ab)|V\rangle,
$$

and vector addition satisfies

$$
|V\rangle + (|W\rangle + |Z\rangle)
=
(|V\rangle + |W\rangle) + |Z\rangle.
$$
<!-- block:end -->

<!-- block:start axiom label=vector-space-identity -->
### Identity

There exists a null vector

$$
|0\rangle
$$

such that

$$
|V\rangle + |0\rangle
=
|V\rangle,
$$

and scalar multiplication satisfies

$$
1|V\rangle
=
|V\rangle.
$$
<!-- block:end -->

<!-- block:start axiom label=additive-inverse -->
### Additive Inverse

For every vector

$$
|V\rangle,
$$

there exists a vector

$$
-|V\rangle
$$

such that

$$
|V\rangle + (-|V\rangle)
=
|0\rangle.
$$
<!-- block:end -->

<!-- subsection:start title="Linear Independence and Basis" -->

<!-- block:start definition label=linear-independence -->
### Linear Independence

A set of vectors

$$
|i\rangle
$$

is linearly independent when the only solution to

$$
\sum_i a_i |i\rangle
=
|0\rangle
$$

is

$$
a_i = 0
\qquad
\forall i.
$$
<!-- block:end -->

<!-- block:start definition label=dimension-of-vector-space -->
### Dimension of a Vector Space

A vector space $X$ has dimension $N$ if there exists a maximal set of

$$
N
$$

linearly independent vectors.
<!-- block:end -->

<!-- block:start definition label=basis-of-vector-space -->
### Basis

A basis of a vector space $X$ is a set of linearly independent vectors that spans the entire space.

Any vector

$$
|V\rangle \in X
$$

can be expanded as

$$
|V\rangle
=
\sum_i v_i |i\rangle.
$$
<!-- block:end -->

<!-- subsection:start title="Inner Product Spaces" -->

<!-- block:start definition label=inner-product-space -->
### Inner Product Space

An inner product space is a vector space $X$ equipped with an inner product

$$
\langle V|W\rangle
$$

satisfying:

1. Conjugate symmetry

$$
\langle V|W\rangle
=
\langle W|V\rangle^*
$$

2. Positive definiteness

$$
\langle V|V\rangle \ge 0
$$

with equality iff

$$
|V\rangle = |0\rangle
$$

3. Linearity in the ket

$$
\langle V|
(a|W\rangle+b|Z\rangle)
=
a\langle V|W\rangle
+
b\langle V|Z\rangle.
$$
<!-- block:end -->

<!-- block:start definition label=orthogonal-vectors -->
### Orthogonal Vectors

Two vectors are orthogonal if

$$
\langle V|W\rangle
=
0.
$$
<!-- block:end -->

<!-- block:start definition label=norm-of-vector -->
### Norm of a Vector

The norm of a vector is defined by

$$
|V|
=
\sqrt{\langle V|V\rangle}.
$$
<!-- block:end -->

<!-- block:start definition label=normalized-vector -->
### Normalized Vector

A vector is normalized when

$$
|V| = 1.
$$
<!-- block:end -->

<!-- block:start definition label=orthonormal-basis -->
### Orthonormal Basis

An orthonormal basis is a complete set of vectors satisfying

$$
\langle n|m\rangle
=
\delta_{nm}.
$$
<!-- block:end -->

<!-- subsection:start title="Dual Spaces and Hermitian Conjugation" -->

<!-- block:start definition label=dual-space -->
### Dual Space

The dual space $X^*$ consists of linear functionals acting on vectors of $X$.

Vectors in the dual space are denoted by bras

$$
\langle V|.
$$
<!-- block:end -->

<!-- block:start definition label=hermitian-conjugate-of-ket -->
### Hermitian Conjugate of a Ket

The Hermitian conjugate operation maps a ket to a bra:

$$
(|V\rangle)^\dagger
=
\langle V|.
$$
<!-- block:end -->

<!-- block:start definition label=hermitian-conjugate-of-bra -->
### Hermitian Conjugate of a Bra

The Hermitian conjugate operation maps a bra to a ket:

$$
(\langle V|)^\dagger
=
|V\rangle.
$$
<!-- block:end -->

<!-- block:start proposition label=double-hermitian-conjugation -->
### Double Hermitian Conjugation

For every ket,

$$
(|V\rangle)^{\dagger\dagger}
=
|V\rangle.
$$
<!-- block:end -->

<!-- block:start proposition label=antilinearity-hermitian-conjugation -->
### Antilinearity of Hermitian Conjugation

For complex scalars $a,b$,

$$
(a|V\rangle+b|W\rangle)^\dagger
=
a^* \langle V|
+
b^* \langle W|.
$$
<!-- block:end -->

<!-- subsection:start title="Gram–Schmidt Orthogonalization" -->

<!-- block:start definition label=gram-schmidt-procedure -->
### Gram–Schmidt Procedure

Given linearly independent vectors

$$
|I\rangle, |II\rangle, |III\rangle,\dots,
$$

an orthonormal basis can be constructed recursively by subtracting projections and normalizing.
<!-- block:end -->

<!-- block:start theorem label=schwarz-inequality -->
### Schwarz Inequality

For any vectors

$$
|V\rangle,
|W\rangle,
$$

one has

$$
|\langle V|W\rangle|^2
\le
\langle V|V\rangle
\langle W|W\rangle.
$$

<!-- proof:start -->
Define

$$
|Z\rangle
=
|V\rangle
-
\frac{\langle W|V\rangle}{\langle W|W\rangle}
|W\rangle.
$$

Since

$$
\langle Z|Z\rangle \ge 0,
$$

expanding the inner product yields

$$
\langle V|V\rangle
-
\frac{
|\langle V|W\rangle|^2
}{
\langle W|W\rangle
}
\ge 0.
$$

Multiplying by

$$
\langle W|W\rangle
$$

gives

$$
|\langle V|W\rangle|^2
\le
\langle V|V\rangle
\langle W|W\rangle.
$$
<!-- proof:end -->
<!-- block:end -->
