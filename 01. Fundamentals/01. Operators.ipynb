<!-- section:start title="Linear and Antilinear Operators" -->

<!-- block:start definition label=operator -->
### Operator

An operator is a mapping

$$
\Omega : X \to X
$$

such that

$$
\Omega |V\rangle
=
|V'\rangle.
$$
<!-- block:end -->

<!-- block:start definition label=linear-operator -->
### Linear Operator

An operator $L$ is linear if

$$
L(\alpha|V\rangle+\beta|W\rangle)
=
\alpha L|V\rangle
+
\beta L|W\rangle.
$$
<!-- block:end -->

<!-- block:start definition label=antilinear-operator -->
### Antilinear Operator

An operator $A$ is antilinear if

$$
A(\alpha|V\rangle+\beta|W\rangle)
=
\alpha^* A|V\rangle
+
\beta^* A|W\rangle.
$$
<!-- block:end -->

<!-- section:start title="Operator Algebra" -->

<!-- block:start definition label=operator-commutator -->
### Commutator

The commutator of two operators is defined by

$$
[\Lambda,\Omega]
=
\Lambda\Omega
-
\Omega\Lambda.
$$
<!-- block:end -->

<!-- block:start definition label=operator-anticommutator -->
### Anticommutator

The anticommutator of two operators is defined by

$$
\{\Lambda,\Omega\}
=
\Lambda\Omega
+
\Omega\Lambda.
$$
<!-- block:end -->

<!-- block:start definition label=outer-product -->
### Outer Product

Given vectors

$$
|V\rangle,
|W\rangle,
$$

the outer product is the operator

$$
|V\rangle\langle W|
$$

defined by

$$
(|V\rangle\langle W|)|\psi\rangle
=
|V\rangle \langle W|\psi\rangle.
$$
<!-- block:end -->

<!-- block:start proposition label=resolution-of-identity -->
### Resolution of the Identity

For an orthonormal basis

$$
|n\rangle,
$$

the identity operator satisfies

$$
\mathbb I
=
\sum_n |n\rangle \langle n|.
$$
<!-- block:end -->

<!-- section:start title="Hermitian and Unitary Operators" -->

<!-- block:start definition label=hermitian-conjugate-operator -->
### Hermitian Conjugate of an Operator

The Hermitian conjugate of an operator $\Omega$ is defined by

$$
\langle V|\Omega^\dagger|W\rangle
=
\langle W|\Omega|V\rangle^*.
$$
<!-- block:end -->

<!-- block:start definition label=hermitian-operator -->
### Hermitian Operator

An operator is Hermitian if

$$
\Omega^\dagger
=
\Omega.
$$
<!-- block:end -->

<!-- block:start definition label=antihermitian-operator -->
### Anti-Hermitian Operator

An operator is anti-Hermitian if

$$
\Omega^\dagger
=
-\Omega.
$$
<!-- block:end -->

<!-- block:start definition label=unitary-operator -->
### Unitary Operator

An operator is unitary if

$$
\Omega\Omega^\dagger
=
\Omega^\dagger\Omega
=
\mathbb I.
$$

Equivalently,

$$
\Omega^{-1}
=
\Omega^\dagger.
$$
<!-- block:end -->

<!-- section:start title="Spectral Theory" -->

<!-- block:start definition label=eigenket -->
### Eigenket

A ket

$$
|u\rangle
$$

is an eigenket of an operator $\Omega$ with eigenvalue $a$ if

$$
\Omega |u\rangle
=
a |u\rangle.
$$
<!-- block:end -->

<!-- block:start definition label=spectrum-of-operator -->
### Spectrum of an Operator

The spectrum of an operator is the set of all its eigenvalues.
<!-- block:end -->

<!-- block:start theorem label=hermitian-operator-real-eigenvalues -->
### Hermitian Operators Have Real Eigenvalues

Let

$$
\Omega^\dagger = \Omega
$$

and suppose

$$
\Omega |u\rangle = a |u\rangle.
$$

Then the eigenvalue

$$
a
$$

is real.

<!-- proof:start -->
Starting from the eigenvalue equation,

$$
\Omega |u\rangle = a |u\rangle,
$$

take the expectation value with respect to

$$
\langle u|.
$$

Then

$$
\langle u|\Omega|u\rangle
=
a \langle u|u\rangle.
$$

Since $\Omega$ is Hermitian,

$$
\langle u|\Omega|u\rangle
=
\langle u|\Omega|u\rangle^*.
$$

Therefore,

$$
a\langle u|u\rangle
=
a^*\langle u|u\rangle.
$$

Hence,

$$
(a-a^*)\langle u|u\rangle = 0.
$$

Because

$$
\langle u|u\rangle > 0,
$$

it follows that

$$
a=a^*.
$$

Therefore the eigenvalue is real.
<!-- proof:end -->
<!-- block:end -->

<!-- block:start theorem label=orthogonality-of-hermitian-eigenspaces -->
### Orthogonality of Eigenspaces of Hermitian Operators

Let

$$
\Omega^\dagger = \Omega
$$

and suppose

$$
\Omega |u\rangle = a |u\rangle,
$$

$$
\Omega |u'\rangle = a' |u'\rangle,
$$

with

$$
a \ne a'.
$$

Then

$$
\langle u'|u\rangle = 0.
$$

<!-- proof:start -->
Using the first eigenvalue equation,

$$
\langle u'|\Omega|u\rangle
=
a\langle u'|u\rangle.
$$

Using the second eigenvalue equation,

$$
\langle u|\Omega|u'\rangle
=
a'\langle u|u'\rangle.
$$

Since $\Omega$ is Hermitian,

$$
\langle u|\Omega|u'\rangle
=
\langle u'|\Omega|u\rangle^*.
$$

Therefore,

$$
a'\langle u|u'\rangle
=
a^*\langle u|u'\rangle.
$$

Because the eigenvalues of Hermitian operators are real,

$$
a^* = a.
$$

Hence,

$$
(a-a')\langle u'|u\rangle = 0.
$$

Since

$$
a \ne a',
$$

it follows that

$$
\langle u'|u\rangle = 0.
$$

Therefore the eigenvectors are orthogonal.
<!-- proof:end -->
<!-- block:end -->
